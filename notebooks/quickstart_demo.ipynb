## 1. Non-Equilibrium Wilson Line Crossing

```python
physics = WetSteamNucleationPhysics()
p_test = 20000.0 # 20 kPa
T_test = 330.0   # 56.85 C
S, r_star = physics.supersaturation_and_critical_radius(p_test, T_test)
J = physics.nucleation_rate(p_test, T_test)
print(f"Supersaturation Ratio S: {S:.2f}")
print(f"Kelvin Critical Droplet Radius r*: {r_star * 1e9:.2f} nm")
print(f"Becker-Doring Nucleation Rate J: {J:.2e} droplets/m3*s")
```
